```python

import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics.pairwise import cosine_similarity

```

```python
# Step 1: Create dataset (hypothetical values)
data = {
    "Building": ["ESIME", "ESIQIE", "ESFM"],
    "Rooms": [45, 35, 40],
    "Floors": [4, 3, 4],
    "Parking_Area": [600, 450, 550],
    "Staircases": [5, 4, 5],
    "Canteens": [2, 1, 1]
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df.to_string(index=False))

```

```python
# Step 2: Extract numerical features
X = df.drop(columns=["Building"])

```

```python
 #Step 3: Normalize features to [0, 1]
scaler = MinMaxScaler()
X_normalized = scaler.fit_transform(X)

normalized_df = pd.DataFrame(
    X_normalized,
    columns=X.columns,
    index=df["Building"]
)

print("\nNormalized Features:")
print(normalized_df.round(3))

```

```python
# Step 4: Calculate cosine similarity
similarity_matrix = cosine_similarity(X_normalized)

result = pd.DataFrame(
    similarity_matrix,
    index=df["Building"],
    columns=df["Building"]
)

print("\nCosine Similarity Matrix:")
print(result.round(4))

```

```python
# Step 5: Compare all building pairs
pairs = []

for i in range(len(df)):
    for j in range(i + 1, len(df)):
        pairs.append((
            df.loc[i, "Building"],
            df.loc[j, "Building"],
            similarity_matrix[i, j]
        ))

print("\nPairwise Similarity:")
for b1, b2, score in pairs:
    print(f"{b1} vs {b2}: {score:.4f}")


```

```python
# Step 6: Identify the most similar pair
most_similar = max(pairs, key=lambda p: p[2])

print("\nMost Similar Buildings:")
print(
    f"{most_similar[0]} and {most_similar[1]} "
    f"(Similarity: {most_similar[2]:.4f})"
)
```
